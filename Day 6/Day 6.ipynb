# BFS


```python
from collections import deque
def bfs(graph, start, goal):
    q = deque([start])
    parent={start:None}
    while q:
        node=q.popleft()
        if node==goal:
            path=[]
            while node is not None:
                path.append(node)
                node=parent[node]
            return path[::-1]
        for nei in graph[node]:
            if nei not in parent:
                parent[nei]=node
                q.append(nei)
    return[] # no path found
graph={
    'A' : ['B','C'],
    'B' : ['D','E'],
    'C' : ['F'],
    'D' : [],
    'E' : ['G'],
    'F' : [],
    'G' : []
}
print("Path:", bfs(graph, 'A', 'G'))
```
