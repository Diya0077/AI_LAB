# A* Search In Python


```python
import heapq # Always gives the smallest element at the front
def astar (graph, start,goal,h):
    openpq = [(0,start)]
    g={ 
        start:0
      }
    parent = {
        start: None
            }
    while openpq:
        f,node = heapq.heappop(openpq) #priority queue stores value in tuple (f,node) 
        if node == goal:
            path = []
            while node is not None:
                path.append(node)
                node = parent[node]
            return path [:: -1]
        for neighbor, cost in graph [node]:
            new_g = g[node]+cost
            # if neighbor is already in the g return current cost if not then neighbor is not yet discover
            if new_g < g.get(neighbor, float ('inf')):
                g[neighbor] = new_g
                parent[neighbor] = node
                f_new = new_g + h( neighbor, goal)
                heapq.heappush(openpq, (f_new,neighbor))
    return []
graph = {
        'A': [('B', 1),('C', 3)],
        'B': [('D', 1),('E', 5)],
        'C': [('F', 2)],
        'D': [('G', 1)],
        'E': [('G', 5)],
        'F': [('G', 2)],
        'G': []
    }
    # heuristic function (Straight line estimation in goal)
h = lambda n, goal:0
print ("Path:", astar(graph, 'A', 'F', h))
```
